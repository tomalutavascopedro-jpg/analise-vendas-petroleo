```python
import plotly.express as px

Meses = [
    "Abril", "Maio", "Junho", "Agosto", 
    "Setembro", "Outubro", "Novembro", "Dezembro"
]

Vendas = [14,18,25,35,44,58,78,90]

Produtos = [
    "Produto A", "Produto B", "Produto C", "Produto D", 
    "Produto E", "Produto F", "Produto H", "Produto I"
]

Names = [
    "Geladeira", "Computador", "Iphones", "Televisão", 
    "Arcas", "Perfumes", "Pastas de escolas", "Cama"
]

região = [
    "Benguela", "Bengo", "Luanda", "Malanje", 
    "Huila", "Menonge", "Huambo", "Zaire"
]

fig = px.bar (
    x = Meses,
    y = Vendas,
    color = região,
    text= Names,
    title="Análise de dados de evolução das Mercadoria"
)
fig.update_xaxes(title="Meses")
fig.update_yaxes(title="Vendas")

fig.show()
```
